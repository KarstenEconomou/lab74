# Calculated characteristics of an n⁺–p–p⁺ silicon solar cell


## DATA

Every value plotted here was computed from the numerical model of J. R. Hauser
and P. M. Dunbar, *A Theoretical Analysis of the Current-Voltage Characteristics
of Solar Cells*, NASA-CR-138828, June 1974
([NTRS 19740020431](https://ntrs.nasa.gov/citations/19740020431)). Nothing is
traced or interpolated from the report's own figures.

The [solver](../data/silicon_solar_cell/hauser_dunbar_1974.py),
[generator](../data/silicon_solar_cell/generate.py), and generated `plot_data.npz`
live together under `data/silicon_solar_cell/`. The solver handles Poisson's equation with both carrier continuity equations in
the normalized electrostatic-potential and quasi-Fermi-level variables of the
report's Table 1, discretized with Scharfetter–Gummel carrier fluxes on a graded
mesh and driven by a damped Newton iteration with a sparse three-colored
Jacobian.

### Baseline cell

The report calculates one $n^+$–$p$–$p^+$ structure in detail. Its Table I and
Table III give:

| Quantity | Value |
| --- | --- |
| Overall thickness | 250 µm |
| $n^+$ emitter width | 0.25 µm, Gaussian diffused |
| $p$ base width | 249.25 µm |
| $p^+$ back-surface field width | 0.5 µm |
| $n^+$ surface concentration | 1 × 10²⁰ cm⁻³ |
| $p$ base doping (10 Ω·cm) | 1.3 × 10¹⁵ cm⁻³ |
| $p^+$ doping | 1 × 10¹⁸ cm⁻³ |
| $\tau_{\mathrm{n0}}$ | 100 µs |
| $\tau_{\mathrm{p0}}$ | 0.1 µs |
| Contacts | ohmic at both faces, i.e. infinite surface recombination |

Recombination is the single-level Shockley–Read–Hall model of the report's
Eq. (17) with the recombination level at midgap. Mobility follows its Eqs. (18)
and (19) after Gwyn et al., with the constants of its Table II
($C_1 = 81$, $C_2 = 2.16 × 10^8$, $C_3 = 1.76 × 10^{-6}$, $C_4 = 1.70 × 10^4$, $C_5 = 1.25 × 10^4$;
$K_1 = 350$, $K_2 = 7.0 × 10^8$, $K_3 = 1.93 × 10^{-6}$, $K_4 = 1.76 × 10^4$, $K_5 = 1.18 × 10^4$),
read from the scanned table rather than from its OCR text, which corrupts $C_2$.
The report does not state a temperature; 300 K is implied by its low-field
mobilities of 1351 and 480 cm² V⁻¹ s⁻¹ and by its remark that the base diffusion
length is "about 588 µM", which those values reproduce.

### AM0 optical generation

Generation follows the report's Eq. (20), developed in its Appendix B, for a
bare polished surface with no antireflection layer and with surface reflection
taken into account:

$$G_{\mathrm{e}}(x) = \int T(\lambda)\,N_{\mathrm{ph}}(\lambda)\,\eta(\lambda)\,
\alpha(\lambda)\,e^{-\alpha(\lambda)x}\,\mathrm{d}\lambda$$

with $T = 1 - R$, $R = ((n-1)^2 + k^2)/((n+1)^2 + k^2)$, a band edge of 1.11 eV, and
the quantum yield of its Eq. (4) after Shockley, $\eta = 3 - 2\exp(3.3 - h\nu)$ above
3.3 eV. The spectral irradiance is Table V of NASA SP-8005 (Thekaekara et al.,
1971), the report's own reference 18, for a solar constant of 135.3 mW cm⁻².
The silicon optical constants are Green's 2008 self-consistent tabulation, which
stands in for the 1972 Philipp data the report used and which is not available
in machine-readable form.

Appendix B, Table 1 of the report publishes four numbers for this case. The
reconstruction reproduces them:

| Quantity | Report | Here |
| --- | ---: | ---: |
| Photons above 1.11 eV | 3.31 × 10¹⁷ cm⁻² s⁻¹ | 3.30 × 10¹⁷ |
| Available current | 52.9 mA cm⁻² | 52.9 |
| Surface loss, bare silicon | 36.4 % | 35.0 |
| Total optical current | 34.2 mA cm⁻² | 34.4 |
| Surface generation rate | 1.15 × 10²² cm⁻³ s⁻¹ | 1.38 × 10²² |

### Orientation, sign, and reference

$x = 0$ is the illuminated $n^+$ surface and $x$ increases toward the rear contact.
The report's own figures use the opposite orientation, placing the illuminated
surface at $x = 250$ µm.

Currents keep the sign of the calculation: forward diode current is positive, so
an illuminated cell delivering power carries negative current. The report's
Figure 17 plots the magnitude of the same quantity.

Panel (d) references both quasi-Fermi levels and the electrostatic energy $-q\psi$
to the Fermi level of the illuminated front contact, so $F_n = F_p = 0$ there
and both fall to $-qV$ at the rear contact.

### How the result compares

The dark characteristic reproduces the behaviour the report describes on its
page 8: the current varies close to $\exp(qV/kT)$ between about 0.35 V and
0.55 V, and more slowly below and above. The calculated open-circuit voltage
agrees with the report's Figure 17 to a few millivolts. The calculated
short-circuit current is about 16 % below the 34.5 mA cm⁻² of that figure. Two
things account for most of that gap, and neither is adjusted for here. The
report states on its page 9 that "the calculated generation rate used in the
presented data was about 10% high" relative to its own Appendix B table. And
Green's ultraviolet absorption coefficients are larger than Philipp's, which
moves generation into the first few nanometres, where the report's infinite
front-surface recombination destroys it.


```python
from pathlib import Path

import numpy as np

repo = Path("..") if Path.cwd().name == "examples" else Path(".")
solar = np.load(repo / "data/silicon_solar_cell/plot_data.npz")
```

## PLOT


```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

import lab74

lab74.use(
    accent=None,
    line_series="ink",
    annotation_size_offset=-2,
    annotation_face="gothic",
)
out = repo / "examples/output"
out.mkdir(exist_ok=True)

BOUNDARIES = solar["boundaries"]
REGIONS = [r"$n^+$ EMITTER", r"$p$ BASE", r"$p^+$ BACK-SURFACE FIELD"]
DEPTH = r"DISTANCE FROM FRONT SURFACE, $x$ (µm)"
SERIES_WIDTH = 1.0  # Keep data distinct from the ink grid.

fig, axes = plt.subplots(2, 2, figsize=(7.0, 7.0))
fig.subplots_adjust(left=0.095, right=0.98, bottom=0.085, top=0.885)
lab74.format_multipanel(axes, grid="ink", hspace=0.27, wspace=0.30)
lab74.panel_labels(
    axes,
    loc="upper right",
    background=True,
    face="mono",
    fontsize=plt.rcParams["axes.labelsize"],
)
lab74.title(
    fig,
    r"CALCULATED CHARACTERISTICS OF AN $n^+$–$p$–$p^+$ SILICON SOLAR CELL",
    "AM0 SOLAR ILLUMINATION (HAUSER AND DUNBAR, 1974)",
)
current_voltage, concentration, components, potentials = axes.flat

# (a) terminal characteristics
current_voltage.plot(
    solar["am0_voltage"],
    solar["am0_current"],
    marker="",
    linewidth=SERIES_WIDTH,
    label="AM0",
)
current_voltage.plot(
    solar["dark_voltage"],
    solar["dark_current"],
    marker="",
    linewidth=SERIES_WIDTH,
    label="DARK",
)
current_voltage.set(
    title="CURRENT–VOLTAGE CHARACTERISTICS",
    xlabel=r"CELL VOLTAGE, $V$ (V)",
    ylabel=r"CURRENT DENSITY, $J$ (mA cm$^{-2}$)",
    xlim=(0, 0.7),
    ylim=(-40, 40),
)
current_voltage.xaxis.set_major_locator(mticker.MultipleLocator(0.1))
current_voltage.xaxis.set_minor_locator(mticker.MultipleLocator(0.05))
current_voltage.yaxis.set_major_locator(mticker.MultipleLocator(10))
current_voltage.yaxis.set_minor_locator(mticker.MultipleLocator(5))
lab74.marked_point(
    current_voltage,
    0.0,
    solar["short_circuit_current"],
    r"$J_{\mathrm{SC}}$",
    offset=(5, 5),
    background=True,
    leader=True,
)
lab74.marked_point(
    current_voltage,
    solar["open_circuit_voltage"],
    0.0,
    r"$V_{\mathrm{OC}}$",
    offset=(4, 0),
    background=True,
)
lab74.marked_point(
    current_voltage,
    solar["maximum_power_voltage"],
    solar["maximum_power_current"],
    "MAXIMUM POWER",
    offset=(24, -8),
    ha="right",
    background=True,
)
lab74.legend(current_voltage, loc="upper left", background=True)

# (b) electron concentration
for condition, name in (
    ("short_circuit", r"SHORT CIRCUIT, $V = 0$"),
    ("maximum_power", r"MAXIMUM POWER, $V = V_{\mathrm{MP}}$"),
    ("open_circuit", r"OPEN CIRCUIT, $V = V_{\mathrm{OC}}$"),
):
    concentration.plot(
        solar[f"{condition}_x"],
        solar[f"{condition}_electrons"],
        marker="",
        linewidth=SERIES_WIDTH,
        label=name,
    )
concentration.set(
    title="ELECTRON CONCENTRATION",
    xlabel=DEPTH,
    ylabel=r"ELECTRON CONCENTRATION, $n$ (cm$^{-3}$)",
    xscale="log",
    yscale="log",
    xlim=(1e-3, BOUNDARIES[-1]),
    ylim=(1e2, 1e21),
)
concentration.yaxis.set_major_locator(mticker.LogLocator(numticks=8))
lab74.legend(concentration, loc="center left", background=True)
lab74.region_labels(
    concentration,
    BOUNDARIES,
    REGIONS,
    position=0.05,
    background=True,
    leader_offset=(8, 8),
)

# (c) current components at maximum power
for values, name in (
    (solar["maximum_power_electron_current"], r"ELECTRON CURRENT, $J_n$"),
    (solar["maximum_power_hole_current"], r"HOLE CURRENT, $J_p$"),
    (
        solar["maximum_power_electron_current"] + solar["maximum_power_hole_current"],
        r"TOTAL CURRENT, $J$",
    ),
):
    components.plot(
        solar["maximum_power_x"],
        values,
        marker="",
        linewidth=SERIES_WIDTH,
        label=name,
    )
components.set(
    title="CURRENT COMPONENTS AT MAXIMUM POWER",
    xlabel=DEPTH,
    ylabel="CURRENT DENSITY (mA cm$^{-2}$)",
    xscale="log",
    xlim=(1e-3, BOUNDARIES[-1]),
    ylim=(-36, 14),
)
components.yaxis.set_major_locator(mticker.MultipleLocator(10))
components.yaxis.set_minor_locator(mticker.MultipleLocator(5))
lab74.legend(
    components, loc="upper left", title=r"AM0 $V = V_{\mathrm{MP}}$", background=True
)
lab74.region_labels(
    components,
    BOUNDARIES,
    REGIONS,
    position=0.05,
    background=True,
    leader_offset=(8, 8),
)

# (d) potentials at maximum power
for values, name in (
    (-solar["maximum_power_potential"], r"ELECTROSTATIC ENERGY, $-q\psi$"),
    (-solar["maximum_power_electron_level"], r"ELECTRON QUASI-FERMI LEVEL, $F_n$"),
    (-solar["maximum_power_hole_level"], r"HOLE QUASI-FERMI LEVEL, $F_p$"),
):
    potentials.plot(
        solar["maximum_power_x"],
        values,
        marker="",
        linewidth=SERIES_WIDTH,
        label=name,
    )
potentials.set(
    title="POTENTIALS AT MAXIMUM POWER",
    xlabel=DEPTH,
    ylabel="ENERGY RELATIVE TO REFERENCE (eV)",
    xscale="log",
    xlim=(1e-3, BOUNDARIES[-1]),
    ylim=(-0.8, 0.2),
)
potentials.yaxis.set_major_locator(mticker.MultipleLocator(0.2))
potentials.yaxis.set_minor_locator(mticker.MultipleLocator(0.1))
lab74.legend(
    potentials,
    loc="upper left",
    title=r"AM0 $V = V_{\mathrm{MP}}$",
    background=True,
)
lab74.region_labels(
    potentials,
    BOUNDARIES,
    REGIONS,
    position=0.05,
    background=True,
    leader_offset=(8, 8),
)

for ax in (concentration, components, potentials):
    ax.xaxis.set_major_locator(mticker.LogLocator(numticks=7))

fig.savefig(out / "04_silicon_solar_cell.png")
plt.close(fig)
```
