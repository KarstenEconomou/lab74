# APOLLO 17 SOIL GRAIN SIZE


## DATA

Weight percent held in each sieve class for Apollo 17 soil 74220,6 (orange soil) and 74240,61 (gray soil), collected at Station 4 on the rim of Shorty Crater, from John C. Graf's NASA *Lunar Soils Grain Size Catalog*, [NASA-RP-1265](https://ntrs.nasa.gov/citations/19930012474). The values below reproduce the corresponding files in NASA's [machine-readable catalog digitization](https://ntrs.nasa.gov/citations/20250001791) by Wesley A. Chambers and Anthony M. DeStefano. Both splits use the same sieve openings and preserve the original laboratory classes.

The plot converts sieve diameter only to the catalog's phi scale, $\phi=-\log_2(d/\mathrm{mm})$. Each reported weight is drawn between its sieve opening and the preceding coarser opening. The leading zero at 10 mm supplies the first class edge and is therefore not plotted as a separate interval. Values are neither interpolated nor renormalized.

```python
from io import StringIO
import numpy as np

records = {
    "74220 — ORANGE SOIL": """diameter_um,weight_percent
10000.0,0.00
4000.0,10.78
2000.0,1.87
1000.0,0.88
500.0,0.00
250.0,4.23
150.0,6.66
90.0,11.45
75.0,4.44
45.0,12.76
20.0,22.71
15.6,7.94
7.8,9.64
3.9,4.56
1.0,1.07
""",
    "74240 — GRAY SOIL": """diameter_um,weight_percent
10000.0,0.00
4000.0,5.73
2000.0,7.22
1000.0,5.87
500.0,4.90
250.0,8.02
150.0,7.34
90.0,10.18
75.0,3.63
45.0,10.78
20.0,15.57
15.6,2.57
7.8,10.32
3.9,6.00
1.0,0.87
""",
}
data = {
    name: np.genfromtxt(StringIO(values), delimiter=",", names=True)
    for name, values in records.items()
}
```

## PLOT


```python
from pathlib import Path
import matplotlib.pyplot as plt
import lab74

lab74.use()
out = Path("output")
out.mkdir(exist_ok=True)
fig, ax = plt.subplots()

orange = data["74220 — ORANGE SOIL"]
gray = data["74240 — GRAY SOIL"]
phi_edges = -np.log2(orange["diameter_um"] / 1000)
lab74.stairs(
    ax,
    orange["weight_percent"][1:],
    phi_edges,
    label="74220 ORANGE SOIL",
)
lab74.stairs(
    ax,
    gray["weight_percent"][1:],
    phi_edges,
    color=lab74.INK,
    fill=False,
    hatch="",
    label="74240 GRAY SOIL",
)
ax.set(
    xlim=(-3.35, 10),
    ylim=(0, 25),
    xlabel=r"PARTICLE DIAMETER, $\phi=-\log_2(d/\mathrm{mm})$",
    ylabel="WEIGHT PERCENT PER SIEVE CLASS",
)
lab74.legend(ax)

fig.savefig(out / "03_apollo17_soil_grain_size.png")
plt.close(fig)
```
