# APOLLO 17 SOIL GRAIN SIZE


## DATA

Weight percent held in each sieve class for Apollo 17 soil 74220,6 (orange soil) and 74240,61 (gray soil), collected at Station 4 on the rim of Shorty Crater, from John C. Graf's NASA *Lunar Soils Grain Size Catalog*, [NASA-RP-1265](https://ntrs.nasa.gov/citations/19930012474). The local files for [74220](../data/apollo17_soil_grain_size/74220_orange_soil.csv) and [74240](../data/apollo17_soil_grain_size/74240_gray_soil.csv) reproduce the corresponding files in NASA's [machine-readable catalog digitization](https://ntrs.nasa.gov/citations/20250001791) by Wesley A. Chambers and Anthony M. DeStefano. Both splits use the same sieve openings and preserve the original laboratory classes.

The plot converts sieve diameter only to the catalog's phi scale, $\phi=-\log_2(d/\mathrm{mm})$. Each reported weight is drawn between its sieve opening and the preceding coarser opening. The leading zero at 10 mm supplies the first class edge and is therefore not plotted as a separate interval. Values are neither interpolated nor renormalized.

```python
from pathlib import Path

import numpy as np

repo = Path("..") if Path.cwd().name == "examples" else Path(".")
data_dir = repo / "data/apollo17_soil_grain_size"
data = {
    "74220 — ORANGE SOIL": np.genfromtxt(
        data_dir / "74220_orange_soil.csv", delimiter=",", names=True
    ),
    "74240 — GRAY SOIL": np.genfromtxt(
        data_dir / "74240_gray_soil.csv", delimiter=",", names=True
    ),
}
```

## PLOT


```python
import matplotlib.pyplot as plt
import lab74

lab74.use("instrument")
out = repo / "examples/output"
out.mkdir(exist_ok=True)
fig, ax = plt.subplots()

orange = data["74220 — ORANGE SOIL"]
gray = data["74240 — GRAY SOIL"]
phi_edges = -np.log2(orange["diameter_um"] / 1000)
lab74.stairs(
    ax,
    orange["weight_percent"][1:],
    phi_edges,
    label="74220 ORANGE SOIL",
)
lab74.stairs(
    ax,
    gray["weight_percent"][1:],
    phi_edges,
    color=lab74.INK,
    fill=False,
    hatch="",
    label="74240 GRAY SOIL",
)
ax.set(
    xlim=(-3.35, 10),
    ylim=(0, 25),
    xlabel=r"PARTICLE DIAMETER, $\phi=-\log_2(d/\mathrm{mm})$",
    ylabel="WEIGHT PERCENT PER SIEVE CLASS",
)
lab74.legend(ax)

fig.savefig(out / "03_apollo17_soil_grain_size.png")
plt.close(fig)
```
