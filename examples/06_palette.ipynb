# lab74 colour reproduction standard

A technical-print reference sheet demonstrating `lab74.table` with text and swatch cells. The descriptive labels are presentation metadata; the public palette contains only each accent name and hexadecimal value.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import lab74

lab74.use(accent=None, face="mono")
out = Path("examples/output") if Path("examples").exists() else Path("output")
out.mkdir(exist_ok=True)

spot_metadata = {
    "instrument": ("Orange", "Fermilab"),
    "aerospace": ("NASA Red", "NASA, 1976"),
    "oxide": ("Massey Ferguson Red", "Fermilab"),
    "research": ("NAL Green", "Fermilab"),
    "survey": ("Prairie Gold Dark", "Fermilab"),
    "telemetry": ("Marlin Blue Dark", "Fermilab"),
    "technology": ("Technology Purple", "EPA, 1977"),
    "laboratory": ("NAL Blue", "FNAL, 1970s"),
}

basic_rows = [
    [
        "PAPER",
        "FIGURE GROUND",
        lab74.TableSwatch(lab74.PAPER),
    ],
    ["INK", "TYPE / RULES / DATA", lab74.TableSwatch(lab74.INK)],
    ["RULE", "CONSTRUCTION", lab74.TableSwatch(lab74.RULE)],
]
spot_rows = [
    [
        name.upper(),
        spot_metadata[name][0].upper(),
        spot_metadata[name][1].upper(),
        lab74.TableSwatch(hex_value),
    ]
    for name, hex_value in lab74.ACCENTS.items()
]

fig, ax = plt.subplots(figsize=(6.5, 4.75))
fig.subplots_adjust(left=0.065, right=0.985, bottom=0.025, top=0.985)
ax.set_axis_off()
page_inset = 0.005
lab74.header(
    ax,
    "LAB74",
    "COLOUR REPRODUCTION STANDARD",
    "REV. 1.0.0",
    inset=page_inset,
)

lab74.table(
    ax,
    basic_rows,
    columns=["NAME", "USE", "SPECIMEN"],
    title="BASIC MATERIALS",
    column_widths=[1.25, 2.25, 2.00],
    bbox=(0, 0.6631, 1, 0.255),
    inset=page_inset,
)
lab74.table(
    ax,
    spot_rows,
    columns=["NAME", "COLOUR", "SOURCE", "SPECIMEN"],
    title="ACCENT COLOUR — USE ONE ONLY",
    column_widths=[1.25, 1.25, 1.00, 2.00],
    bbox=(0, 0.016, 1, 0.624),
    inset=page_inset,
)

fig.savefig(out / "06_palette.png")
plt.close(fig)
```
