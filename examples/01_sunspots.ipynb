# 1974 SUNSPOT ACTIVITY


## DATA

Daily total sunspot number, version 2.0, from the [WDC-SILSO, Royal Observatory of Belgium](https://www.sidc.be/SILSO/DATA/SN_d_tot_V2.0.csv), DOI [10.24414/qnza-ac80](https://doi.org/10.24414/qnza-ac80). The local [plot extract](../data/sunspots/silso_daily_total_1974_weekly.csv) retains January 1 and every seventh daily record thereafter (53 of 365 definitive 1974 observations); values are neither averaged nor interpolated. `sigma` is SILSO's reported daily standard-deviation field. SILSO notes that pre-1981 errors are estimates from an autoregressive model based on the Poisson distribution, with the observation count set to one. Data license: [CC BY-NC 4.0](https://creativecommons.org/licenses/by-nc/4.0/).


```python
from pathlib import Path

import numpy as np

repo = Path("..") if Path.cwd().name == "examples" else Path(".")
data = np.genfromtxt(
    repo / "data/sunspots/silso_daily_total_1974_weekly.csv",
    delimiter=",",
    names=True,
    dtype=[("date", "datetime64[D]"), ("sunspots", "f8"), ("sigma", "f8")],
)
```

## PLOT


```python
from datetime import datetime
import matplotlib.dates as mdates
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import lab74

lab74.use(accent=None)
out = repo / "examples/output"
out.mkdir(exist_ok=True)
fig, ax = plt.subplots()
lab74.errorbar(ax, data["date"], data["sunspots"], yerr=data["sigma"])
ax.set(
    xlim=(datetime(1973, 12, 28), datetime(1974, 12, 31, 23, 59)),
    ylim=(0, 160),
    xlabel="CALENDAR MONTH",
    ylabel=r"SUNSPOT NUMBER, $R$",
)
ax.xaxis.set_major_locator(mdates.MonthLocator(bymonth=range(1, 13, 2)))
ax.xaxis.set_major_formatter(
    mticker.FuncFormatter(
        lambda value, _: mdates.num2date(value).strftime("%b").upper()
    )
)
ax.xaxis.set_minor_locator(mdates.MonthLocator())
ax.yaxis.set_major_locator(mticker.MultipleLocator(20))
ax.yaxis.set_minor_locator(mticker.MultipleLocator(10))
lab74.format_frame(ax, style="open")

fig.savefig(out / "01_sunspots.png")
plt.close(fig)
```
